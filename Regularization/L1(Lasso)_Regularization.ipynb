```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Lasso
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import SGDRegressor
```

# ***L2 REGULARIZATION CLASS FOR SIMPLE LINEAR REGRESSION DATA(2D)***

```python
X,y = make_regression(n_samples=350, n_features=1, n_informative=1, n_targets=1,noise=39,random_state=26)
```

```python
plt.scatter(X,y, color='Orange')
```

***Scikit-learn's Linear Regression class***

```python
lr_model = LinearRegression()
lr_model.fit(X,y)
print(f'Linear Regression Class Results: b = {lr_model.intercept_} and m = {lr_model.coef_}')
```

***Scikit-learn's Lasso Regularization class***

```python
lasso_model = Lasso(alpha=0.1)
lasso_model.fit(X,y)
print(f'Ridge Regression Class Results: b = {lasso_model.intercept_} and m = {lasso_model.coef_}')
```
